# 01 — Data Cleaning

Loads the five raw source files, standardises them, and writes tidy tables to `data/cleaned/`.

| Input (`data/raw/`) | Meaning |
|---|---|
| `01_people.csv` | one row per person |
| `03_education.csv` | education history |
| `04_experience.csv` | employment history |
| `05_person_skills.csv` | person → skill (long format) |
| `coursera_data.csv` | the course catalogue |

**Run order:** `01_data_cleaning` → `02_recommendation_data` → `03_recommendation_engine`.

> This notebook needs the raw files. If you only have the prepared outputs in
> `data/recommendation_ready/`, skip to `02`, which can rebuild the cleaned tables from them.

```python
from pathlib import Path

import numpy as np
import pandas as pd

# Resolve paths relative to the notebook so the pipeline runs from any working directory.
PROJECT_ROOT = Path.cwd()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
CLEAN_DIR = PROJECT_ROOT / "data" / "cleaned"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

SOURCES = {
    "people": RAW_DIR / "01_people.csv",
    "education": RAW_DIR / "03_education.csv",
    "experience": RAW_DIR / "04_experience.csv",
    "person_skills": RAW_DIR / "05_person_skills.csv",
    "courses": RAW_DIR / "coursera_data.csv",
}

missing = [name for name, path in SOURCES.items() if not path.exists()]
if missing:
    raise FileNotFoundError(
        f"Missing raw files in {RAW_DIR}: {missing}.\n"
        "Place the raw CSVs there, or skip to 02_recommendation_data.ipynb, "
        "which can rebuild the cleaned tables from data/recommendation_ready/."
    )

print(f"Raw data directory: {RAW_DIR}")
```

```python
def clean_column_names(df):
    df = df.copy()
    
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )
    
    return df


people = clean_column_names(people)
education = clean_column_names(education)
experience = clean_column_names(experience)
person_skills = clean_column_names(person_skills)
courses = clean_column_names(courses)
```

```python
people = people[["person_id", "name"]].copy()

people["person_id"] = pd.to_numeric(
    people["person_id"],
    errors="coerce"
).astype("Int64")

people["name"] = people["name"].apply(clean_text)

# Remove rows without an ID
people = people.dropna(subset=["person_id"])

# person_id should uniquely identify a person
people = people.drop_duplicates(subset=["person_id"])

print(people.shape)
print(people.head())
```

```python
education = education[
    ["person_id", "institution", "program", "location"]
].copy()

education["person_id"] = pd.to_numeric(
    education["person_id"],
    errors="coerce"
).astype("Int64")

for col in ["institution", "program", "location"]:
    education[col] = education[col].apply(clean_text)

education = education.dropna(subset=["person_id"])

# Remove completely duplicated education records
education = education.drop_duplicates()

print(education.shape)
print(education.head())
```

```python
experience = experience[
    ["person_id", "title", "firm", "location"]
].copy()

experience["person_id"] = pd.to_numeric(
    experience["person_id"],
    errors="coerce"
).astype("Int64")

for col in ["title", "firm", "location"]:
    experience[col] = experience[col].apply(clean_text)

experience = experience.dropna(subset=["person_id"])

experience = experience.drop_duplicates()

print(experience.shape)
print(experience.head())
```

```python
person_skills = person_skills[
    ["person_id", "skill"]
].copy()

person_skills["person_id"] = pd.to_numeric(
    person_skills["person_id"],
    errors="coerce"
).astype("Int64")

person_skills["skill"] = person_skills["skill"].apply(clean_text)

person_skills = person_skills.dropna(
    subset=["person_id", "skill"]
)

# Normalize skill matching
person_skills["skill_normalized"] = (
    person_skills["skill"]
    .str.lower()
    .str.strip()
)

# Remove exact duplicate person-skill relationships
person_skills = person_skills.drop_duplicates(
    subset=["person_id", "skill_normalized"]
)

print(person_skills.shape)
print(person_skills.head())
```

### Duplicate-person check

`person_id` is unique, so `drop_duplicates(subset=["person_id"])` cannot detect a person who was
loaded more than once under different ids. We therefore fingerprint each person by their *content*
(name + sorted skill set + job titles) and count how often that fingerprint repeats.

A repeat factor above 1.0 means the source file contains stacked copies of the same population —
which inflates every frequency statistic downstream and leaks across any train/test split.

```python
def person_fingerprint(people, person_skills, experience):
    """Content-based identity for a person, independent of person_id."""
    skills = (
        person_skills.groupby("person_id")["skill_normalized"]
        .agg(lambda s: "|".join(sorted(set(s))))
        .rename("skill_sig")
    )
    titles = (
        experience.groupby("person_id")["title"]
        .agg(lambda s: "|".join(sorted({str(v) for v in s})))
        .rename("title_sig")
    )
    fp = people.set_index("person_id").join([skills, titles])
    return (
        fp["name"].fillna("~").astype(str)
        + "||" + fp["skill_sig"].fillna("~")
        + "||" + fp["title_sig"].fillna("~")
    )


fingerprints = person_fingerprint(people, person_skills, experience)
distinct_people = fingerprints.nunique()
repeat_factor = len(fingerprints) / distinct_people

print(f"person_id rows      : {len(fingerprints):,}")
print(f"distinct people     : {distinct_people:,}")
print(f"repeat factor       : {repeat_factor:.2f}x")

if repeat_factor > 1.05:
    print(
        f"\nWARNING: the source population appears to be stacked ~{repeat_factor:.0f}x.\n"
        "         02_recommendation_data.ipynb de-duplicates on this fingerprint.\n"
        "         Do not trust raw frequency counts from this notebook."
    )
```
